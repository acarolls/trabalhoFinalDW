## Tema 1 - Notificações de casos de dengue no Estado do Rio de Janeiro


```python
# Importando bibliotecas
import pandas as pd
import numpy as np
import matplotlib as plt
```

```python
# Importando o dataset 
df_2022 = pd.read_csv('data/DengueRJ2022.csv', sep=',')
df_2023 = pd.read_csv('data/DengueRJ2023.csv', sep=',', dtype={'DT_CHIK_S1': str, 'DT_PRNT': str})
df_2024 = pd.read_csv('data/DengueRJ2024.csv', sep=',', dtype={'DT_CHIK_S1': str, 'DT_PRNT': str})
df_2025 = pd.read_csv('data/DengueRJ2025.csv', sep=',')
df_2026 = pd.read_csv('data/DengueRJ2026.csv', sep=',')

df_2023['DT_CHIK_S1'] = pd.to_datetime(
    df_2023['DT_CHIK_S1'], errors='coerce'
)
df_2024['DT_CHIK_S1'] = pd.to_datetime(
    df_2024['DT_CHIK_S1'], errors='coerce'
)

df_2023['DT_PRNT'] = pd.to_datetime(
    df_2023['DT_PRNT'], errors='coerce'
)

df_2024['DT_PRNT'] = pd.to_datetime(
    df_2024['DT_PRNT'], errors='coerce'
)
```

```python
print("Tamanho do dataset de 2022: ", df_2022.shape)
df_2022.describe()
```

```python
print("Tamanho do dataset de 2023: ", df_2023.shape)
df_2023.describe()
```

```python
print("Tamanho do dataset de 2024: ", df_2024.shape)
df_2024.describe()
```

```python
print("Tamanho do dataset de 2025: ", df_2025.shape)
df_2025.describe()
```

```python
print("Tamanho do dataset de 2026: ", df_2026.shape)
df_2026.describe()
```

```python
# Junção dos datasets de acordo com as colunas em comum
df_joined = pd.concat([df_2022, df_2023, df_2024, df_2025, df_2026], ignore_index=True)
print("Tamanho do dataset unido: ", df_joined.shape)
df_joined.describe()
```

```python
# Calculando a porcentagem de células vazias por coluna no df
empty_percentage = (df_joined.isnull().sum() / len(df_joined)) * 100
removed_columns = []

empty_percentage_col = {
    col: float(percentage)
    for col, percentage in empty_percentage.items()
}

for col, percentage in empty_percentage_col.items(): 
    if percentage >= 90.0:
        removed_columns.append(col)
        df_joined.drop(col, axis=1, inplace=True)
        


print(empty_percentage_col)
print('\nQuantidade de colunas removidas com 90% ou mais das células vazias:', len(empty_percentage_col) - df_joined.shape[1])
print('Colunas removidas:', removed_columns)
print('Quantidade de colunas restantes:', df_joined.shape[1])
print('Quantidade de colunas com 0% das células vazias: ', len([col for col, percentage in empty_percentage_col.items() if percentage == 0.0]))
print('Colunas restantes: ', df_joined.columns.tolist())
```
